```python
## total revenue 
total_revenue = data['Sales'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
## revenue by city
revenue_by_city = data.groupby('City')['Sales'].sum()
print(f"Highest Revenue by City:{revenue_by_city.idxmax()}: ${revenue_by_city.max():,.2f}")
```

```python
## profit by branch
profit_by_branch = data.groupby('Branch')['Sales'].sum()
print(f"Most Profitable Branch:{profit_by_branch.idxmax()}: ${profit_by_branch.max():,.2f}")
```

```python
## profit and revenue by product category
profit_by_product = data.groupby('Product_line')['Sales'].sum()
print(f"Top Performing Category:{profit_by_product.idxmax()}: ${profit_by_product.max():,.2f}")
```

```python
## spending by customer type
spending_by_customer_type = data.groupby('Customer type')['Sales'].sum()
print(f"Customer that spent the most:{spending_by_customer_type.idxmax()}: ${spending_by_customer_type.max():,.2f}")
```

```python
## payment method
payment_method = data.groupby('Payment')['Sales'].sum()
print(f"Most Popular Payment Method:{payment_method.idxmax()}: ${payment_method.max():,.2f}")
```

```python
## average transaction value
average_transaction_value = data['Sales'].mean()
print(f"Average Transaction Value: ${average_transaction_value:,.2f}")
```

```python
## customer satisfacation by branch
satisfaction_by_branch = data.groupby('Branch')['satisfied'].value_counts(normalize=True).unstack().fillna(0)
highest_branch = satisfaction_by_branch['satisfied'].idxmax()
highest_satisfaction = satisfaction_by_branch['satisfied'].max()

print(f"Branch with Highest Satisfaction:"f"{highest_branch}:{highest_satisfaction * 100:.2f}%")
```
